```python
import pandas as pd
```

```python
df = pd.read_excel(r'C:\Офисы\Дашборд по офисам\The deals_All data.xlsx', sheet_name='Base_сделки')
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip()
```

```python
df['Дата публикации'] = pd.to_datetime(
    df['Дата публикации'],
    format='%d.%m.%Y'
)
```

```python
df['Дата публикации'].unique()
```

```python
columns_to_keep = [
    '#',
    'Кв',
    'Год',
    'Консультант',
    'Type of deal (обмен)',
    'Тип консультации для сводной',
    'Тип сделки для сводной',
    'Общ. S сделки, кв. м',
    'S сделки, кв. м',
    'Для расчета среднего',
    'Размер сделки',
    'Целиком/блок',
    'S, кв. м для обсчета',
    'Кол-во р/м, шт.',
    'shell',
    'БЦ rus / eng',
    'Адрес',
    'Координаты',
    'Класс',
    'Зона Москвы',
    'Компания',
    'Компания dl',
    'rus/for',
    'Бизнес сектор для обсчета',
    'в коворкинг',
    'T-rep',
    'LLR/ETR',
    'quartercalculation',
    'Дата публикации'
]

# Если в названиях столбцов могут быть лишние пробелы
df.columns = df.columns.str.strip()

# Оставить только нужные столбцы
df = df[columns_to_keep]
```

```python
df['quartercalculation'] = df['Дата публикации'].dt.strftime('%d.%m.')
```

```python
df['quartercalculation'].unique()
```

```python
import csv

df.to_csv(
    r'C:\Офисы\Дашборд по офисам\The_deals_All_data_cleaned.csv',
    index=False,
    sep=",",
    encoding="utf-8",
    quoting=csv.QUOTE_MINIMAL
)
```
