```python
import re

def normalize_address(addr):
    if pd.isna(addr):
        return addr

    s = str(addr).strip()

    # убрать переносы строк и двойные пробелы
    s = s.replace('\r', ' ').replace('\n', ' ')
    s = re.sub(r'\s+', ' ', s)

    # заменить разные тире
    s = s.replace('–', '-').replace('—', '-')

    # словарь замен (без учета регистра)
    replacements = {
        # английские
        r'\bhgh\b': 'highway',
        r'\bhwy\b': 'highway',
        r'\bsh\.\b': 'shosse',
        r'\bsh\b': 'shosse',
        r'\bul\.\b': 'ulitsa',
        r'\bul\b': 'ulitsa',
        r'\bpr-t\b': 'prospekt',
        r'\bpr\.\b': 'prospekt',
        r'\bpr\b': 'prospekt',
        r'\bper\.\b': 'pereulok',
        r'\bper\b': 'pereulok',
        r'\bnab\.\b': 'embankment',
        r'\bnab\b': 'embankment',
        r'\bemb\.\b': 'embankment',
        r'\bemb\b': 'embankment',
        r'\bbldg\.\b': 'building',
        r'\bbldg\b': 'building',
        r'\bbld\.\b': 'building',
        r'\bbld\b': 'building',
        r'\bstr\.\b': 'stroenie',
        r'\bstr\b': 'stroenie',

        # русские
        r'\bпр-т\b': 'проспект',
        r'\bпр\.\b': 'проспект',
        r'\bпер\.\b': 'переулок',
        r'\bнаб\.\b': 'набережная',
        r'\bул\.\b': 'улица',
        r'\bд\.\b': '',
        r'\bкорп\.\b': 'корпус',
        r'\bкорп\b': 'корпус',
        r'\bк\.\b': 'корпус',
        r'\bстр\.\b': 'строение',
        r'\bстр\b': 'строение',
    }

    for pattern, repl in replacements.items():
        s = re.sub(pattern, repl, s, flags=re.IGNORECASE)

    # building1 -> building 1
    s = re.sub(r'building\s*([0-9]+)', r'building \1', s, flags=re.IGNORECASE)

    # stroenie1 -> stroenie 1
    s = re.sub(r'stroenie\s*([0-9]+)', r'stroenie \1', s, flags=re.IGNORECASE)

    # корпус2 -> корпус 2
    s = re.sub(r'корпус\s*([0-9]+)', r'корпус \1', s)

    # строение2 -> строение 2
    s = re.sub(r'строение\s*([0-9]+)', r'строение \1', s)

    # убрать повторные запятые и пробелы
    s = re.sub(r'\s*,\s*', ', ', s)
    s = re.sub(r'\s+', ' ', s).strip(' ,')

    # добавить Moscow только если его нет
    if ('москва' not in s.lower()) and ('moscow' not in s.lower()):
        s += ', Moscow'

    return s
```

```python
import re
import pandas as pd

def normalize_bc_name(name):
    if pd.isna(name):
        return name

    s = str(name)

    # Переводы строк, табы, неразрывные пробелы
    s = (
        s.replace('\r', ' ')
         .replace('\n', ' ')
         .replace('\t', ' ')
         .replace('\xa0', ' ')
    )

    # Разные тире
    s = s.replace('–', '-').replace('—', '-')

    # Кавычки
    s = (
        s.replace('«', '"')
         .replace('»', '"')
         .replace('“', '"')
         .replace('”', '"')
         .replace("’", "'")
         .replace("`", "'")
    )

    # Один пробел
    s = re.sub(r'\s+', ' ', s)

    # Убрать пробелы вокруг /
    s = re.sub(r'\s*/\s*', '/', s)

    # Убрать пробелы вокруг дефиса
    s = re.sub(r'\s*-\s*', '-', s)

    # Привести к нижнему регистру
    s = s.casefold()

    return s.strip()
```

```python
# Маска строк с заполненным названием БЦ
mask = df['Адрес_норм'].notna() & (df['БЦ rus / eng'].str.strip() != '')
```

```python
import numpy as np

df['Столбец1'] = df['Столбец1'].replace('', np.nan)

df.loc[mask, 'Столбец1'] = (
    df.loc[mask]
      .groupby('БЦ rus / eng')['Столбец1']
      .transform(lambda x: x.ffill().bfill())
)
```

```python
import numpy as np

df['Latitude'] = df['Latitude'].replace('', np.nan)

df.loc[mask, 'Latitude'] = (
    df.loc[mask]
      .groupby('БЦ rus / eng')['Latitude']
      .transform(lambda x: x.ffill().bfill())
)
```

```python
# Пустые строки -> NaN
df['Longitude'] = df['Longitude'].replace('', np.nan)

# Заполняем только их
df.loc[mask, 'Longitude'] = (
    df.loc[mask]
      .groupby('БЦ rus / eng')['Longitude']
      .transform(lambda x: x.ffill().bfill())
)
```
